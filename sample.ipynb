## amex

```python
import json

json_path = '/data/sfs/home/rensiyu/Attention/datasets/GUI-Actor-Data/amex/amex_bbox.json'
with open(json_path, 'r', encoding='utf-8') as f:
    data = json.load(f)
print(data[0])
print(f"len(data):{len(data)}")
```

```python
import os
# import pandas as pd
from tqdm import tqdm

# 一个数据文件记录了所有图片路径
def check_missing_images(dataset_path, data):
    missing_files = []
    
    # 读取数据文件（根据你的实际格式调整）
    # data = pd.read_json(data_file) 或其他格式
    
    # 使用tqdm添加进度条，显示处理进度
    for item in tqdm(data, desc="检查图片文件", unit="files"):
        image_path = item.get('image')
        image_path = os.path.join(dataset_path, image_path)
        if not os.path.exists(image_path):
            missing_files.append(image_path)
    
    print(f"\n找到 {len(missing_files)} 个缺失文件")
    # for file in missing_files[:10]:  # 只显示前10个
    #     print(f"缺失: {file}")
    
    return missing_files

# 使用示例
check_missing_images(dataset_path='/data/sfs/home/rensiyu/Attention/datasets/GUI-Actor-Data/amex/screenshots', data=data)
```

```python
print(data[0]['conversations'][1::2])
b = set()
for i in range(len(data)):
    temp = data[i]['conversations'][1::2]
    for j in range(len(temp)):
        b.add(temp[j]['from'])  # 逐个添加

print(b)
```

```python
a = set()
for i in range(len(data)):
    if data[i]['conversations'] is None:
        print('data[i]')
        break
    a.add(len(data[i]['conversations']))
print(a)
print(f"len(set):{len(a)}")
```
